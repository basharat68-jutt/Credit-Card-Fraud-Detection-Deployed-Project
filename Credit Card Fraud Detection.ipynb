```python
"""Feature Engeering and Preprocessing"""

from sklearn.preprocessing import OneHotEncoder,StandardScaler

ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore').set_output(transform='pandas')

merchant_ohe = ohe.fit_transform(df[['merchant_category']])


scaler = StandardScaler().set_output(transform='pandas')
scaler_columns = df.drop(columns=['transaction_id','merchant_category','is_fraud'])
scaled_columns = scaler.fit_transform(scaler_columns)

new_df = np.concatenate((scaled_columns,merchant_ohe),axis=1)

```

# Train-Test-Split

```python
from sklearn.model_selection import train_test_split

x = new_df
y = df['is_fraud']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

# LogisticRegression Prediction

```python
from sklearn.linear_model import LogisticRegression

lo = LogisticRegression()

lo.fit(x_train,y_train)
y_pred = lo.predict(x_test)

print(y_pred)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score

print(f"Accuracy Score: {accuracy_score(y_test,y_pred)}")
print(f"Precision Score: {precision_score(y_test,y_pred)}")
print(f"Recall Score: {recall_score(y_test,y_pred)}")
```

# RandomForrestClassifier

```python
from sklearn.ensemble import RandomForestClassifier

tree = RandomForestClassifier()

tree.fit(x_train,y_train)
tree_pred = tree.predict(x_test)

print(f"Accuracy Score: {accuracy_score(y_test,tree_pred)}")
print(f"Precision Score: {precision_score(y_test,tree_pred)}")
print(f"Recall Score: {recall_score(y_test,tree_pred)}")
```

# Hyperparameter Tunning

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

params = {
    "n_estimators" : [50,100,200],
    'max_depth' : [None,5,10,20],
    'max_features' : ['sqrt','log2'],
    'min_samples_split' : [2,4,10],
    'min_samples_leaf' : [1,2,4],
    'class_weight' : [None, 'balanced']
    }

grid_search = GridSearchCV(
    estimator=tree,
    param_grid=params,
    cv=5,
    scoring='f1',
    n_jobs=-1,
    verbose=1
)

grid_search.fit(x_train, y_train)

print("Best Parameters", grid_search.best_params_)
print("Best CV Score", grid_search.best_score_)

best_model = grid_search.best_estimator_


```

# Piepline

```python
print(df.columns)
```

```python
from sklearn.compose import ColumnTransformer

str_col = ['merchant_category']
numeric_col = ['amount', 'transaction_hour',
       'foreign_transaction', 'location_mismatch', 'device_trust_score',
       'velocity_last_24h', 'cardholder_age']

feature_col = numeric_col + str_col
target_col = ['is_fraud']

x = df[feature_col]
y = df[target_col]

xp_train, xp_test, yp_train, yp_test = train_test_split(x,y,random_state=42,test_size=0.2)


trf_1 = ColumnTransformer([
    ('ohe',OneHotEncoder(sparse_output=False,handle_unknown='ignore'),str_col)
],remainder='passthrough',
verbose_feature_names_out=False).set_output(transform='pandas')

trf_2 = ColumnTransformer([('scaler',StandardScaler(),numeric_col)
                           ],remainder='passthrough',verbose_feature_names_out=False).set_output(transform='pandas')

trf_3 = RandomForestClassifier(random_state=42)

from sklearn.pipeline import Pipeline

rf_pipeline = Pipeline([
    ('trf_1',trf_1),
    ('trf_2',trf_2),
    ('trf_3',trf_3)
])
```

```python
rf_pipeline.fit(xp_train, yp_train)
pipe_pred = rf_pipeline.predict(xp_test)

print('Accuracy_score',accuracy_score(y_test,pipe_pred))
print('Recall Score',recall_score(y_test,pipe_pred))
```

```python
import joblib
joblib.dump(rf_pipeline,"credit_card_fraud_detection.pkl")
```
